```python
# ==========================================
# Hybrid Model 1 : KNN + Random Forest
# ==========================================

from sklearn.ensemble import VotingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    cohen_kappa_score,
    confusion_matrix,
    classification_report
)
import joblib

knn = KNeighborsClassifier(n_neighbors=5)

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

ensemble_knn_rf = VotingClassifier(
    estimators=[
        ("knn", knn),
        ("rf", rf)
    ],
    voting="soft"
)

# Train
ensemble_knn_rf.fit(X_train_scaled, y_train)

# Predict
y_pred = ensemble_knn_rf.predict(X_test_scaled)
y_prob = ensemble_knn_rf.predict_proba(X_test_scaled)[:, 1]

# Metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_prob)
kappa = cohen_kappa_score(y_test, y_pred)

print("=" * 50)
print("Hybrid KNN + RF Results")
print("=" * 50)

print(f"Accuracy      : {accuracy:.4f}")
print(f"Precision     : {precision:.4f}")
print(f"Recall        : {recall:.4f}")
print(f"F1 Score      : {f1:.4f}")
print(f"ROC-AUC Score : {auc:.4f}")
print(f"Cohen Kappa   : {kappa:.4f}")

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))

joblib.dump(ensemble_knn_rf, "../model/knn_rf_ensemble.pkl")

print("\nHybrid Model Saved Successfully!")
```

```python
# ==========================================
# Hybrid Model 2 : Logistic Regression + Random Forest
# ==========================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    cohen_kappa_score,
    confusion_matrix,
    classification_report
)
import joblib

# Base Models
lr = LogisticRegression(random_state=42)

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

# Hybrid Model
ensemble_lr_rf = VotingClassifier(
    estimators=[
        ("lr", lr),
        ("rf", rf)
    ],
    voting="soft"
)

# Train
ensemble_lr_rf.fit(X_train_scaled, y_train)

# Predict
y_pred_lr_rf = ensemble_lr_rf.predict(X_test_scaled)
y_prob_lr_rf = ensemble_lr_rf.predict_proba(X_test_scaled)[:,1]

# Metrics
accuracy_lr_rf = accuracy_score(y_test, y_pred_lr_rf)
precision_lr_rf = precision_score(y_test, y_pred_lr_rf)
recall_lr_rf = recall_score(y_test, y_pred_lr_rf)
f1_lr_rf = f1_score(y_test, y_pred_lr_rf)
auc_lr_rf = roc_auc_score(y_test, y_prob_lr_rf)
kappa_lr_rf = cohen_kappa_score(y_test, y_pred_lr_rf)

print("="*50)
print("Hybrid Logistic Regression + RF Results")
print("="*50)

print(f"Accuracy      : {accuracy_lr_rf:.4f}")
print(f"Precision     : {precision_lr_rf:.4f}")
print(f"Recall        : {recall_lr_rf:.4f}")
print(f"F1 Score      : {f1_lr_rf:.4f}")
print(f"ROC-AUC Score : {auc_lr_rf:.4f}")
print(f"Cohen Kappa   : {kappa_lr_rf:.4f}")

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_lr_rf))

print("\nClassification Report")
print(classification_report(y_test, y_pred_lr_rf))

joblib.dump(ensemble_lr_rf, "../model/lr_rf_ensemble.pkl")

print("\nHybrid LR + RF Model Saved Successfully!")
```

```python
# ==========================================
# Hybrid Model 3 : Decision Tree + Random Forest
# ==========================================

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    cohen_kappa_score,
    confusion_matrix,
    classification_report
)
import joblib

# Base Models
dt = DecisionTreeClassifier(random_state=42)

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

# Hybrid Model
ensemble_dt_rf = VotingClassifier(
    estimators=[
        ("dt", dt),
        ("rf", rf)
    ],
    voting="soft"
)

# Train
ensemble_dt_rf.fit(X_train, y_train)

# Predict
y_pred_dt_rf = ensemble_dt_rf.predict(X_test)
y_prob_dt_rf = ensemble_dt_rf.predict_proba(X_test)[:,1]

# Evaluation
accuracy_dt_rf = accuracy_score(y_test, y_pred_dt_rf)
precision_dt_rf = precision_score(y_test, y_pred_dt_rf)
recall_dt_rf = recall_score(y_test, y_pred_dt_rf)
f1_dt_rf = f1_score(y_test, y_pred_dt_rf)
auc_dt_rf = roc_auc_score(y_test, y_prob_dt_rf)
kappa_dt_rf = cohen_kappa_score(y_test, y_pred_dt_rf)

print("="*50)
print("Hybrid Decision Tree + RF Results")
print("="*50)

print(f"Accuracy      : {accuracy_dt_rf:.4f}")
print(f"Precision     : {precision_dt_rf:.4f}")
print(f"Recall        : {recall_dt_rf:.4f}")
print(f"F1 Score      : {f1_dt_rf:.4f}")
print(f"ROC-AUC Score : {auc_dt_rf:.4f}")
print(f"Cohen Kappa   : {kappa_dt_rf:.4f}")

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_dt_rf))

print("\nClassification Report")
print(classification_report(y_test, y_pred_dt_rf))

joblib.dump(ensemble_dt_rf, "../model/dt_rf_ensemble.pkl")

print("\nHybrid DT + RF Model Saved Successfully!")
```
